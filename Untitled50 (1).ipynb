```python

import os
import time
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
import tensorflow_datasets as tfds

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# Experiment settings
IMG_SIZE = 224
BATCH_SIZE = 16
EPOCHS = 10
NUM_CLASSES = 5

AUTOTUNE = tf.data.AUTOTUNE

print("TensorFlow version:", tf.__version__)
print("GPU devices:", tf.config.list_physical_devices("GPU"))
print("Epochs:", EPOCHS)
```

```python

(ds_all, ds_info) = tfds.load(
    "tf_flowers",
    split="train",
    as_supervised=True,
    with_info=True,
    shuffle_files=False
)

class_names = ds_info.features["label"].names

images = []
labels = []

for image, label in tfds.as_numpy(ds_all):
    images.append(image)
    labels.append(int(label))

images = np.array(images, dtype=object)
labels = np.array(labels, dtype=np.int64)

print("Total images:", len(labels))
print("Number of classes:", len(class_names))
print("Classes:", class_names)
```

```python

indices = np.arange(len(labels))

# 70% training and 30% temporary
train_idx, temp_idx = train_test_split(
    indices,
    test_size=0.30,
    random_state=SEED,
    stratify=labels
)

# Split remaining 30% equally into validation and testing
val_idx, test_idx = train_test_split(
    temp_idx,
    test_size=0.50,
    random_state=SEED,
    stratify=labels[temp_idx]
)

print("Training images:", len(train_idx))
print("Validation images:", len(val_idx))
print("Testing images:", len(test_idx))
print(
    "Percentages:",
    round(100 * len(train_idx) / len(labels), 2),
    round(100 * len(val_idx) / len(labels), 2),
    round(100 * len(test_idx) / len(labels), 2)
)
```

```python

def preprocess_image(image, label, training=False):
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = tf.cast(image, tf.float32)

    if training:
        image = tf.image.random_flip_left_right(image)
        image = tf.image.random_brightness(image, max_delta=0.08)

    image = tf.keras.applications.resnet50.preprocess_input(image)
    label = tf.one_hot(tf.cast(label, tf.int32), NUM_CLASSES)

    return image, label


def create_dataset(indices, training=False):
    selected_images = [images[i] for i in indices]
    selected_labels = labels[indices]

    def generator():
        for img, label in zip(selected_images, selected_labels):
            yield img, label

    dataset = tf.data.Dataset.from_generator(
        generator,
        output_signature=(
            tf.TensorSpec(
                shape=(None, None, 3),
                dtype=tf.uint8
            ),
            tf.TensorSpec(shape=(), dtype=tf.int64)
        )
    )

    if training:
        dataset = dataset.shuffle(
            len(indices),
            seed=SEED,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        lambda img, label: preprocess_image(
            img, label, training
        ),
        num_parallel_calls=AUTOTUNE
    )

    return dataset.batch(BATCH_SIZE).prefetch(AUTOTUNE)


train_ds = create_dataset(train_idx, training=True)
val_ds = create_dataset(val_idx, training=False)
test_ds = create_dataset(test_idx, training=False)

print("Training batches:", tf.data.experimental.cardinality(train_ds).numpy())
print("Validation batches:", tf.data.experimental.cardinality(val_ds).numpy())
print("Testing batches:", tf.data.experimental.cardinality(test_ds).numpy())
```

```python

from tensorflow.keras import layers, Model

def build_baseline_model():
    backbone = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_shape=(IMG_SIZE, IMG_SIZE, 3)
    )

    # Freeze pretrained layers for initial transfer learning
    backbone.trainable = False

    inputs = layers.Input(
        shape=(IMG_SIZE, IMG_SIZE, 3)
    )

    features = backbone(inputs, training=False)

    x = layers.GlobalAveragePooling2D()(features)
    x = layers.Dropout(0.3)(x)

    outputs = layers.Dense(
        NUM_CLASSES,
        activation="softmax"
    )(x)

    model = Model(inputs, outputs, name="Baseline_ResNet50")

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=1e-3
        ),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


baseline_model = build_baseline_model()
baseline_model.summary()
```

```python

baseline_history = baseline_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    verbose=1
)
```

```python

from tensorflow.keras import layers

def channel_attention(x, ratio=16, name="channel_attention"):
    channels = int(x.shape[-1])

    # Squeeze: summarize each feature channel
    avg = layers.GlobalAveragePooling2D(
        name=f"{name}_gap"
    )(x)

    # Excitation: learn channel importance
    hidden = layers.Dense(
        max(channels // ratio, 1),
        activation="relu",
        name=f"{name}_fc1"
    )(avg)

    weights = layers.Dense(
        channels,
        activation="sigmoid",
        name=f"{name}_fc2"
    )(hidden)

    weights = layers.Reshape(
        (1, 1, channels),
        name=f"{name}_weights"
    )(weights)

    # Apply channel weights to the feature map
    refined = layers.Multiply(
        name=f"{name}_scale"
    )([x, weights])

    return refined, weights
```

```python

def spatial_attention(x, kernel_size=7, name="spatial_attention"):
    # Average and maximum across channels
    avg = layers.Lambda(
        lambda t: tf.reduce_mean(t, axis=-1, keepdims=True),
        name=f"{name}_avg"
    )(x)

    maximum = layers.Lambda(
        lambda t: tf.reduce_max(t, axis=-1, keepdims=True),
        name=f"{name}_max"
    )(x)

    # Combine both spatial descriptors
    combined = layers.Concatenate(
        axis=-1,
        name=f"{name}_concat"
    )([avg, maximum])

    # Generate spatial attention weights
    weights = layers.Conv2D(
        filters=1,
        kernel_size=kernel_size,
        padding="same",
        activation="sigmoid",
        name=f"{name}_weights"
    )(combined)

    refined = layers.Multiply(
        name=f"{name}_scale"
    )([x, weights])

    return refined, weights
```

```python

def build_attention_model(attention_type="channel"):
    inputs = layers.Input(
        shape=(IMG_SIZE, IMG_SIZE, 3),
        name="image"
    )

    backbone = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_shape=(IMG_SIZE, IMG_SIZE, 3)
    )

    # Extract the pretrained network up to P3.
    # Attention is inserted after conv4_x.
    p3_extractor = tf.keras.Model(
        backbone.input,
        backbone.get_layer("conv4_block6_out").output,
        name="resnet_to_p3"
    )

    p3_extractor.trainable = False
    x = p3_extractor(inputs)

    channel_weights = None
    spatial_weights = None

    if attention_type == "channel":
        x, channel_weights = channel_attention(
            x, name="ca"
        )

    elif attention_type == "spatial":
        x, spatial_weights = spatial_attention(
            x, name="sa"
        )

    elif attention_type == "channel_spatial":
        x, channel_weights = channel_attention(
            x, name="ca"
        )
        x, spatial_weights = spatial_attention(
            x, name="sa"
        )

    # Pass the refined P3 features through the remaining
    # pretrained conv5_x stage, preserving residual blocks.
    # This is constructed in the next cell.
    return inputs, backbone, x, channel_weights, spatial_weights
```

```python

def apply_resnet_stage(backbone, x, stage, blocks):
    """Reapply pretrained ResNet bottleneck blocks to a new input tensor."""
    for block in blocks:
        prefix = f"conv{stage}_block{block}"

        shortcut = x

        # First 1x1 convolution
        y = backbone.get_layer(f"{prefix}_1_conv")(x)
        y = backbone.get_layer(f"{prefix}_1_bn")(y)

        # Second 3x3 convolution
        y = backbone.get_layer(f"{prefix}_2_conv")(y)
        y = backbone.get_layer(f"{prefix}_2_bn")(y)

        # Third 1x1 convolution
        y = backbone.get_layer(f"{prefix}_3_conv")(y)
        y = backbone.get_layer(f"{prefix}_3_bn")(y)

        # Projection shortcut for the first block
        if block == 1:
            shortcut = backbone.get_layer(
                f"{prefix}_0_conv"
            )(shortcut)
            shortcut = backbone.get_layer(
                f"{prefix}_0_bn"
            )(shortcut)

        x = backbone.get_layer(
            f"{prefix}_add"
        )([y, shortcut])

        x = backbone.get_layer(
            f"{prefix}_out"
        )(x)

    return x
```

```python

def build_attention_model(attention_type="channel"):
    inputs = layers.Input(
        shape=(IMG_SIZE, IMG_SIZE, 3),
        name="image"
    )

    backbone = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_shape=(IMG_SIZE, IMG_SIZE, 3)
    )

    # Freeze all pretrained backbone layers
    backbone.trainable = False

    # Extract features through conv4_x (P3)
    p3_extractor = tf.keras.Model(
        backbone.input,
        backbone.get_layer("conv4_block6_out").output
    )

    x = p3_extractor(inputs, training=False)

    channel_weights = None
    spatial_weights = None

    if attention_type == "channel":
        x, channel_weights = channel_attention(
            x, name="ca"
        )

    elif attention_type == "spatial":
        x, spatial_weights = spatial_attention(
            x, name="sa"
        )

    elif attention_type == "channel_spatial":
        x, channel_weights = channel_attention(
            x, name="ca"
        )
        x, spatial_weights = spatial_attention(
            x, name="sa"
        )

    # Continue through conv5_x (P4) with pretrained weights
    x = apply_resnet_stage(
        backbone, x, stage=5, blocks=range(1, 4)
    )

    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.3)(x)

    outputs = layers.Dense(
        NUM_CLASSES, activation="softmax",
        name="classifier"
    )(x)

    model = Model(
        inputs, outputs,
        name=f"ResNet50_{attention_type}"
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model, channel_weights, spatial_weights
```

```python

attention_results = {}
attention_models = {}
attention_histories = {}
attention_maps = {}

for attention_type in [
    "channel",
    "spatial",
    "channel_spatial"
]:
    print("\nTraining:", attention_type)

    model, channel_map, spatial_map = (
        build_attention_model(attention_type)
    )

    history = model.fit(
        train_ds,
        validation_data=val_ds,
        epochs=EPOCHS,
        verbose=1
    )

    attention_models[attention_type] = model
    attention_histories[attention_type] = history

    attention_maps[attention_type] = {
        "channel": channel_map,
        "spatial": spatial_map
    }

print("All three attention models have finished training.")
```

```python

def self_attention(x, name="self_attention"):
    channels = int(x.shape[-1])
    reduced_channels = max(channels // 8, 32)

    # Query, Key and Value projections
    query = layers.Conv2D(
        reduced_channels, 1, name=f"{name}_query"
    )(x)

    key = layers.Conv2D(
        reduced_channels, 1, name=f"{name}_key"
    )(x)

    value = layers.Conv2D(
        channels, 1, name=f"{name}_value"
    )(x)

    # Flatten the spatial dimensions: 14 x 14 = 196
    query = layers.Reshape(
        (-1, reduced_channels)
    )(query)

    key = layers.Reshape(
        (-1, reduced_channels)
    )(key)

    value = layers.Reshape(
        (-1, channels)
    )(value)

    # Scaled dot-product attention
    scores = layers.Lambda(
        lambda t: tf.matmul(
            t[0], t[1], transpose_b=True
        ) / tf.sqrt(
            tf.cast(reduced_channels, tf.float32)
        ),
        name=f"{name}_scores"
    )([query, key])

    attention_weights = layers.Softmax(
        axis=-1, name=f"{name}_weights"
    )(scores)

    attended = layers.Lambda(
        lambda t: tf.matmul(t[0], t[1]),
        name=f"{name}_apply"
    )([attention_weights, value])

    attended = layers.Reshape(
        (int(x.shape[1]), int(x.shape[2]), channels)
    )(attended)

    # Learnable residual scaling
    gamma = self_attention_gamma = tf.Variable(
        0.0, trainable=True, name=f"{name}_gamma"
    )

    output = layers.Lambda(
        lambda t: t[0] + self_attention_gamma * t[1],
        name=f"{name}_residual"
    )([x, attended])

    return output, attention_weights
```

```python

def build_attention_model(attention_type="channel"):
    inputs = layers.Input(
        shape=(IMG_SIZE, IMG_SIZE, 3),
        name="image"
    )

    backbone = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_shape=(IMG_SIZE, IMG_SIZE, 3)
    )
    backbone.trainable = False

    p3_extractor = tf.keras.Model(
        backbone.input,
        backbone.get_layer("conv4_block6_out").output
    )

    x = p3_extractor(inputs, training=False)

    channel_weights = None
    spatial_weights = None
    self_weights = None

    # A1: Channel Attention
    if attention_type == "channel":
        x, channel_weights = channel_attention(x, name="ca")

    # A2: Spatial Attention
    elif attention_type == "spatial":
        x, spatial_weights = spatial_attention(x, name="sa")

    # A3: Channel -> Spatial
    elif attention_type == "channel_spatial":
        x, channel_weights = channel_attention(x, name="ca")
        x, spatial_weights = spatial_attention(x, name="sa")

    # A4: Self-Attention
    elif attention_type == "self":
        x, self_weights = self_attention(x, name="self")

    # A5: Combined Channel -> Spatial -> Self-Attention
    elif attention_type == "combined":
        x, channel_weights = channel_attention(x, name="ca")
        x, spatial_weights = spatial_attention(x, name="sa")
        x, self_weights = self_attention(x, name="self")

    else:
        raise ValueError(
            f"Unknown attention type: {attention_type}"
        )

    # Continue through the pretrained conv5_x stage
    x = apply_resnet_stage(
        backbone, x, stage=5, blocks=range(1, 4)
    )

    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.3)(x)

    outputs = layers.Dense(
        NUM_CLASSES,
        activation="softmax",
        name="classifier"
    )(x)

    model = Model(
        inputs, outputs,
        name=f"ResNet50_{attention_type}"
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=1e-3
        ),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model, channel_weights, spatial_weights, self_weights
```

```python

for attention_type in ["self", "combined"]:
    print(f"\nTraining A4/A5: {attention_type}")

    model, channel_map, spatial_map, self_map = (
        build_attention_model(attention_type)
    )

    history = model.fit(
        train_ds,
        validation_data=val_ds,
        epochs=EPOCHS,
        verbose=1
    )

    attention_models[attention_type] = model
    attention_histories[attention_type] = history

    attention_maps[attention_type] = {
        "channel": channel_map,
        "spatial": spatial_map,
        "self": self_map
    }
```

```python

import time
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Include the baseline and all five attention models
all_models = {"A0_Baseline": baseline_model}

for name, model in attention_models.items():
    all_models[name] = model

results = []

for model_name, model in all_models.items():
    y_true_all = []
    y_pred_all = []
    inference_times = []

    # Evaluate each test batch
    for images_batch, labels_batch in test_ds:
        # Support either integer labels or one-hot labels
        labels_np = labels_batch.numpy()
        if labels_np.ndim > 1 and labels_np.shape[-1] > 1:
            true_labels = np.argmax(labels_np, axis=1)
        else:
            true_labels = labels_np.reshape(-1).astype(int)

        # Warm-up is excluded from timing
        _ = model(images_batch, training=False)

        start = time.perf_counter()
        predictions = model(images_batch, training=False)
        elapsed = time.perf_counter() - start

        pred_labels = np.argmax(predictions.numpy(), axis=1)

        y_true_all.extend(true_labels)
        y_pred_all.extend(pred_labels)

        inference_times.append(
            elapsed / int(images_batch.shape[0]) * 1000
        )

    accuracy = accuracy_score(y_true_all, y_pred_all)
    precision = precision_score(
        y_true_all, y_pred_all, average="macro", zero_division=0
    )
    recall = recall_score(
        y_true_all, y_pred_all, average="macro", zero_division=0
    )
    f1 = f1_score(
        y_true_all, y_pred_all, average="macro", zero_division=0
    )

    total_params = model.count_params()
    trainable_params = sum(
        np.prod(weight.shape)
        for weight in model.trainable_weights
    )

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Macro Precision": precision,
        "Macro Recall": recall,
        "Macro F1": f1,
        "Trainable Parameters": int(trainable_params),
        "Total Parameters": int(total_params),
        "Inference Time (ms/image)": np.mean(inference_times)
    })

results_df = pd.DataFrame(results)

print(results_df.round(4).to_string(index=False))
results_df.to_csv("attention_model_results.csv", index=False)
```

```python
def spatial_attention(x, kernel_size=7, name="spatial_attention"):
    # Correctly wrap TF operations in layers.Lambda for symbolic Keras Tensors
    avg = layers.Lambda(
        lambda t: tf.reduce_mean(t, axis=-1, keepdims=True),
        name=f"{name}_avg"
    )(x)

    maximum = layers.Lambda(
        lambda t: tf.reduce_max(t, axis=-1, keepdims=True),
        name=f"{name}_max"
    )(x)

    # Combine both spatial descriptors
    combined = layers.Concatenate(
        axis=-1,
        name=f"{name}_concat"
    )([avg, maximum])

    # Generate spatial attention weights
    weights = layers.Conv2D(
        filters=1,
        kernel_size=kernel_size,
        padding="same",
        activation="sigmoid",
        name=f"{name}_weights"
    )(combined)

    refined = layers.Multiply(
        name=f"{name}_scale"
    )([x, weights])

    return refined, weights
```

```python

def se_attention(x, ratio=16, name="se_attention"):
    channels = int(x.shape[-1])

    # Squeeze: summarize each channel
    squeezed = layers.GlobalAveragePooling2D(
        name=f"{name}_gap"
    )(x)

    # Excitation: learn channel importance
    hidden = layers.Dense(
        max(channels // ratio, 1),
        activation="relu",
        name=f"{name}_fc1"
    )(squeezed)

    weights = layers.Dense(
        channels,
        activation="sigmoid",
        name=f"{name}_fc2"
    )(hidden)

    weights = layers.Reshape(
        (1, 1, channels),
        name=f"{name}_weights"
    )(weights)

    refined = layers.Multiply(
        name=f"{name}_scale"
    )([x, weights])

    return refined, weights
```

```python

def eca_attention(x, kernel_size=3, name="eca_attention"):
    channels = int(x.shape[-1])

    # Global average pooling: (batch, channels)
    pooled = layers.GlobalAveragePooling2D(
        name=f"{name}_gap"
    )(x)

    # Conv1D operates over neighboring channel descriptors
    channel_sequence = layers.Reshape(
        (channels, 1),
        name=f"{name}_reshape"
    )(pooled)

    channel_weights = layers.Conv1D(
        filters=1,
        kernel_size=kernel_size,
        padding="same",
        use_bias=False,
        name=f"{name}_conv"
    )(channel_sequence)

    channel_weights = layers.Activation(
        "sigmoid",
        name=f"{name}_sigmoid"
    )(channel_weights)

    weights = layers.Reshape(
        (1, 1, channels),
        name=f"{name}_weights"
    )(channel_weights)

    refined = layers.Multiply(
        name=f"{name}_scale"
    )([x, weights])

    return refined, weights
```

```python

def build_attention_model_with_kernel(
    attention_type="channel",
    kernel_size=7
):
    inputs = layers.Input(
        shape=(IMG_SIZE, IMG_SIZE, 3),
        name="image"
    )

    backbone = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_shape=(IMG_SIZE, IMG_SIZE, 3)
    )
    backbone.trainable = False

    p3_extractor = tf.keras.Model(
        backbone.input,
        backbone.get_layer("conv4_block6_out").output
    )

    x = p3_extractor(inputs, training=False)

    channel_weights = None
    spatial_weights = None
    self_weights = None

    # A1: Channel Attention
    if attention_type == "channel":
        x, channel_weights = channel_attention(
            x, name="ca"
        )

    # A2: Spatial Attention
    elif attention_type == "spatial":
        x, spatial_weights = spatial_attention(
            x, kernel_size=kernel_size, name="sa"
        )

    # A3: Channel -> Spatial
    elif attention_type == "channel_spatial":
        x, channel_weights = channel_attention(
            x, name="ca"
        )
        x, spatial_weights = spatial_attention(
            x, kernel_size=kernel_size, name="sa"
        )

    # A4: Self-Attention
    elif attention_type == "self":
        x, self_weights = self_attention(
            x, name="self"
        )

    # A5: Channel -> Spatial -> Self-Attention
    elif attention_type == "combined":
        x, channel_weights = channel_attention(
            x, name="ca"
        )
        x, spatial_weights = spatial_attention(
            x, kernel_size=kernel_size, name="sa"
        )
        x, self_weights = self_attention(
            x, name="self"
        )

    # A6: Squeeze-and-Excitation Attention
    elif attention_type == "se":
        x, channel_weights = se_attention(
            x, name="se"
        )

    # A7: Efficient Channel Attention
    elif attention_type == "eca":
        x, channel_weights = eca_attention(
            x, kernel_size=3, name="eca"
        )

    else:
        raise ValueError(
            f"Unknown attention type: {attention_type}"
        )

    # Continue through the pretrained conv5_x stage
    x = apply_resnet_stage(
        backbone, x, stage=5, blocks=range(1, 4)
    )

    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.3)(x)

    outputs = layers.Dense(
        NUM_CLASSES,
        activation="softmax",
        name="classifier"
    )(x)

    model = Model(
        inputs,
        outputs,
        name=f"ResNet50_{attention_type}_k{kernel_size}"
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=1e-3
        ),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return (
        model,
        channel_weights,
        spatial_weights,
        self_weights
    )
```

```python

# Correctly replay the remaining pretrained ResNet50 residual blocks.
def apply_resnet_stage(backbone, x, stage, blocks):
    for block in blocks:
        prefix = f"conv{stage}_block{block}"

        shortcut = x

        # First bottleneck convolution
        y = backbone.get_layer(f"{prefix}_1_conv")(x)
        y = backbone.get_layer(f"{prefix}_1_bn")(y)
        y = backbone.get_layer(f"{prefix}_1_relu")(y)

        # Second bottleneck convolution
        y = backbone.get_layer(f"{prefix}_2_conv")(y)
        y = backbone.get_layer(f"{prefix}_2_bn")(y)
        y = backbone.get_layer(f"{prefix}_2_relu")(y)

        # Third bottleneck convolution
        y = backbone.get_layer(f"{prefix}_3_conv")(y)
        y = backbone.get_layer(f"{prefix}_3_bn")(y)

        # Projection shortcut for the first block of each stage
        if block == 1:
            shortcut = backbone.get_layer(
                f"{prefix}_0_conv"
            )(shortcut)
            shortcut = backbone.get_layer(
                f"{prefix}_0_bn"
            )(shortcut)

        x = backbone.get_layer(
            f"{prefix}_add"
        )([y, shortcut])

        x = backbone.get_layer(f"{prefix}_out")(x)

    return x


def build_channel_spatial_at_position(position="P2"):
    inputs = layers.Input(
        shape=(IMG_SIZE, IMG_SIZE, 3),
        name="image"
    )

    backbone = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_shape=(IMG_SIZE, IMG_SIZE, 3)
    )
    backbone.trainable = False

    # Select the feature map where attention is inserted.
    position_layers = {
        "P2": "conv3_block4_out",
        "P3": "conv4_block6_out",
        "P4": "conv5_block3_out"
    }

    if position not in position_layers:
        raise ValueError("position must be P2, P3, or P4")

    extractor = tf.keras.Model(
        inputs=backbone.input,
        outputs=backbone.get_layer(
            position_layers[position]
        ).output
    )

    x = extractor(inputs, training=False)

    # Channel attention followed by spatial attention
    x, channel_weights = channel_attention(
        x, name=f"{position.lower()}_ca"
    )

    x, spatial_weights = spatial_attention(
        x, kernel_size=7, name=f"{position.lower()}_sa"
    )

    # Continue through any remaining ResNet stages.
    if position == "P2":
        x = apply_resnet_stage(
            backbone, x, stage=4, blocks=range(1, 7)
        )
        x = apply_resnet_stage(
            backbone, x, stage=5, blocks=range(1, 4)
        )

    elif position == "P3":
        x = apply_resnet_stage(
            backbone, x, stage=5, blocks=range(1, 4)
        )

    # P4 is after conv5_x, so no further residual stage is needed.

    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.3)(x)

    outputs = layers.Dense(
        NUM_CLASSES,
        activation="softmax",
        name="classifier"
    )(x)

    model = Model(
        inputs,
        outputs,
        name=f"ResNet50_ChannelSpatial_{position}"
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model, channel_weights, spatial_weights
```

```python

def find_layer_recursive(model, layer_name):
    try:
        return model.get_layer(layer_name)
    except ValueError:
        for layer in model.layers:
            if isinstance(layer, tf.keras.Model):
                try:
                    return find_layer_recursive(layer, layer_name)
                except ValueError:
                    pass
    raise ValueError(f"Could not find layer: {layer_name}")


def make_gradcam_heatmap(model, input_batch, class_index=None):
    target_layer = find_layer_recursive(
        model, "conv5_block3_out"
    )

    grad_model = tf.keras.Model(
        inputs=model.inputs,
        outputs=[target_layer.output, model.output]
    )

    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(
            input_batch, training=False
        )

        if class_index is None:
            class_index = tf.argmax(
                predictions[0]
            )

        class_score = predictions[:, class_index]

    gradients = tape.gradient(class_score, conv_outputs)

    if gradients is None:
        raise ValueError(
            "Gradients are unavailable for this model graph. "
            "We need to expose the final convolutional feature map "
            "from the model builder."
        )

    # Average gradients over spatial positions
    weights = tf.reduce_mean(
        gradients, axis=(1, 2), keepdims=True
    )

    heatmap = tf.reduce_sum(
        weights * conv_outputs, axis=-1
    )
    heatmap = tf.nn.relu(heatmap)
    heatmap = heatmap[0]

    heatmap = heatmap / (
        tf.reduce_max(heatmap) + tf.keras.backend.epsilon()
    )

    return heatmap.numpy(), int(class_index), predictions.numpy()[0]
```

```python

def build_baseline_gradcam_model(trained_model):
    # Find the nested pretrained ResNet50
    backbone = next(
        layer for layer in trained_model.layers
        if isinstance(layer, tf.keras.Model)
        and "resnet50" in layer.name.lower()
    )

    # Extract the last convolutional feature map from the backbone
    feature_extractor = tf.keras.Model(
        backbone.input,
        backbone.get_layer("conv5_block3_out").output
    )

    # Reconnect the existing classifier head to the feature map
    inputs = layers.Input(
        shape=(IMG_SIZE, IMG_SIZE, 3),
        name="gradcam_image"
    )

    features = feature_extractor(inputs, training=False)

    x = trained_model.layers[-3](features)
    x = trained_model.layers[-2](x)
    predictions = trained_model.layers[-1](x)

    return tf.keras.Model(
        inputs,
        [features, predictions],
        name="baseline_gradcam_model"
    )


baseline_gradcam_model = build_baseline_gradcam_model(
    baseline_model
)

print("Baseline Grad-CAM graph created.")
```

```python

def build_attention_gradcam_model(attention_type="channel_spatial"):
    inputs = layers.Input(
        shape=(IMG_SIZE, IMG_SIZE, 3),
        name="image"
    )

    backbone = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_shape=(IMG_SIZE, IMG_SIZE, 3)
    )
    backbone.trainable = False

    p3_extractor = tf.keras.Model(
        backbone.input,
        backbone.get_layer("conv4_block6_out").output
    )
    p3_extractor.trainable = False

    x = p3_extractor(inputs, training=False)

    if attention_type == "channel":
        x, _ = channel_attention(x, name="ca")

    elif attention_type == "spatial":
        x, _ = spatial_attention(x, kernel_size=7, name="sa")

    elif attention_type == "channel_spatial":
        x, _ = channel_attention(x, name="ca")
        x, _ = spatial_attention(x, kernel_size=7, name="sa")

    else:
        raise ValueError(
            "Use 'channel', 'spatial', or 'channel_spatial'."
        )

    x = apply_resnet_stage(
        backbone, x, stage=5, blocks=range(1, 4)
    )

    # Save this connected feature tensor for Grad-CAM
    final_features = x

    pooled = layers.GlobalAveragePooling2D()(final_features)
    dropped = layers.Dropout(0.3)(pooled)

    predictions = layers.Dense(
        NUM_CLASSES,
        activation="softmax",
        name="classifier"
    )(dropped)

    # Same graph provides both feature maps and predictions
    gradcam_model = Model(
        inputs,
        [final_features, predictions],
        name=f"GradCAM_{attention_type}"
    )

    training_model = Model(inputs, predictions)
    training_model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return training_model, gradcam_model
```

```python

def calculate_gradcam(gradcam_model, input_batch):
    with tf.GradientTape() as tape:
        features, predictions = gradcam_model(
            input_batch, training=False
        )

        class_index = tf.argmax(predictions[0])
        score = predictions[:, class_index]

    gradients = tape.gradient(score, features)

    if gradients is None:
        raise ValueError("Grad-CAM gradients are disconnected.")

    weights = tf.reduce_mean(
        gradients, axis=(1, 2), keepdims=True
    )

    heatmap = tf.reduce_sum(weights * features, axis=-1)
    heatmap = tf.nn.relu(heatmap)[0]

    heatmap /= (
        tf.reduce_max(heatmap) + tf.keras.backend.epsilon()
    )

    return heatmap.numpy(), int(class_index.numpy()), predictions.numpy()[0]


def compare_gradcam_same_image(input_batch, resized_image):
    baseline_map, baseline_class, baseline_probs = (
        calculate_gradcam(baseline_gradcam_model, input_batch)
    )

    attention_map, attention_class, attention_probs = (
        calculate_gradcam(attention_gradcam_model, input_batch)
    )

    baseline_map = tf.image.resize(
        baseline_map[..., None], (IMG_SIZE, IMG_SIZE)
    ).numpy().squeeze()

    attention_map = tf.image.resize(
        attention_map[..., None], (IMG_SIZE, IMG_SIZE)
    ).numpy().squeeze()

    display_image = np.clip(resized_image / 255.0, 0, 1)

    fig, axes = plt.subplots(1, 3, figsize=(16, 5))

    axes[0].imshow(display_image)
    axes[0].set_title("Input Image")

    axes[1].imshow(display_image)
    axes[1].imshow(baseline_map, cmap="jet", alpha=0.45)
    axes[1].set_title(
        f"Baseline Grad-CAM\n{class_names[baseline_class]}"
    )

    axes[2].imshow(display_image)
    axes[2].imshow(attention_map, cmap="jet", alpha=0.45)
    axes[2].set_title(
        f"Channel + Spatial Grad-CAM\n{class_names[attention_class]}"
    )

    for ax in axes:
        ax.axis("off")

    plt.tight_layout()
    plt.show()


compare_gradcam_same_image(input_batch, resized_image)
```

```python

def plot_self_attention(model, self_tensor, input_batch,
                        resized_image, model_name="Self-Attention"):
    attention_model = tf.keras.Model(
        inputs=model.inputs,
        outputs=self_tensor
    )

    attention = attention_model(
        input_batch, training=False
    ).numpy()[0]

    num_positions = attention.shape[-1]
    side = int(np.sqrt(num_positions))

    if side * side != num_positions:
        raise ValueError(
            "Attention positions cannot be reshaped into a square map."
        )

    center_index = num_positions // 2
    response = attention[center_index].reshape(side, side)

    response = tf.image.resize(
        response[..., None], (IMG_SIZE, IMG_SIZE)
    ).numpy().squeeze()

    image_display = np.clip(resized_image / 255.0, 0, 1)

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    axes[0].imshow(image_display)
    axes[0].set_title("Input")
    axes[0].axis("off")

    axes[1].imshow(response, cmap="jet")
    axes[1].set_title("Central Query Response")
    axes[1].axis("off")

    axes[2].imshow(image_display)
    axes[2].imshow(response, cmap="jet", alpha=0.4)
    axes[2].set_title("Self-Attention Overlay")
    axes[2].axis("off")

    plt.suptitle(model_name)
    plt.tight_layout()
    plt.show()


plot_self_attention(
    attention_models["self"],
    attention_maps["self"]["self"],
    input_batch,
    resized_image,
    model_name="A4 Self-Attention"
)
```

```python
import time
import numpy as np
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
# Import safe IPython display function to bypass the shadowed global name 'display'
from IPython.display import display as ipython_display

# Connect model names to models already trained in your notebook.
# Missing models are skipped rather than assigned fabricated values.
comparison_models = {
    "No Attention": baseline_model,
    "Channel / SE": attention_models.get("channel"),
    "Spatial Attention": attention_models.get("spatial"),
    "ECA": attention_models.get("eca"),
    "CBAM": attention_models.get("cbam"),
    "Self-Attention": attention_models.get("self"),
    "Additional Mechanism": attention_models.get("combined")
}

# Collect test images and labels
X_test = []
y_test = []

for xb, yb in test_ds:
    X_test.append(xb.numpy())
    labels = yb.numpy()

    if labels.ndim > 1 and labels.shape[-1] > 1:
        labels = np.argmax(labels, axis=1)

    y_test.extend(labels.reshape(-1).astype(int))

X_test = np.concatenate(X_test, axis=0)
y_test = np.asarray(y_test)

def evaluate_for_table(model, X, y, repeats=3):
    # Warm-up to reduce first-call overhead
    _ = model(X[:1], training=False)

    predictions = model.predict(X, verbose=0)
    y_pred = np.argmax(predictions, axis=1)

    accuracy = accuracy_score(y, y_pred)
    precision = precision_score(
        y, y_pred, average="macro", zero_division=0
    )
    recall = recall_score(
        y, y_pred, average="macro", zero_division=0
    )
    macro_f1 = f1_score(
        y, y_pred, average="macro", zero_division=0
    )

    # Measure average inference latency per image.
    # Use batches and repeat timing for a more stable estimate.
    batch_size = 16
    timing_runs = []

    for _ in range(repeats):
        start = time.perf_counter()

        for start_idx in range(0, len(X), batch_size):
            _ = model(
                X[start_idx:start_idx + batch_size],
                training=False
            )

        elapsed = time.perf_counter() - start
        timing_runs.append(elapsed / len(X) * 1000)

    inference_ms = float(np.median(timing_runs))

    return {
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "Macro F1": macro_f1,
        "Parameters": model.count_params(),
        "Inference Time (ms/image)": inference_ms
    }


comparison_rows = []

for name, model in comparison_models.items():
    if model is None:
        print(f"Skipping {name}: no trained model found.")
        continue

    print(f"Evaluating {name}...")

    try:
        metrics = evaluate_for_table(model, X_test, y_test)
        comparison_rows.append({
            "Attention": name,
            **metrics
        })
    except Exception as e:
        print(f"Could not evaluate {name}: {e}")

comparison_df = pd.DataFrame(comparison_rows)

if not comparison_df.empty:
    # Using the explicitly imported safe ipython_display function to avoid conflicts
    ipython_display(
        comparison_df.style.format({
            "Accuracy": "{:.4f}",
            "Precision": "{:.4f}",
            "Recall": "{:.4f}",
            "Macro F1": "{:.4f}",
            "Parameters": "{:,.0f}",
            "Inference Time (ms/image)": "{:.4f}"
        })
    )
else:
    print("No models were successfully evaluated.")
```
